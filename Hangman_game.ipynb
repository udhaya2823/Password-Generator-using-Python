```python
# Hangman game

import random

word_list = ["ardvark","baboon","camel"]
chosen_word = random.choice(word_list)
word_length = len(chosen_word)

display = []
for i in range(word_length):
  display += "_"
print(display)

end_of_game = False
lives = 6

stages = ['''
              +---+
                |   |
                O   |
               /|\  |
               / \  |
                    |
              =======
               ''','''
              +---+
                |   |
                O   |
               /|\  |
               /    |
                    |
              =======
               ''','''
              +---+
                |   |
                O   |
               /|\  |
                    |
                    |
              =======
               ''','''
              +---+
                |   |
                O   |
               /|   |
                    |
                    |
              =======
               ''','''
              +---+
                |   |
                O   |
                |   |
                    |
                    |
              =======
               ''','''
              +---+
                |   |
                O   |
                    |
                    |
                    |
              =======
               ''','''
              +---+
                |   |
                    |
                    |
                    |
                    |
              ======= ''']

while not end_of_game:
  guess_letter = input("Guess a letter: ").lower()

  if len(guess_letter) > 1:
    print(f"Invalid input '{guess_letter}', Guess and Enter a single letter")
    continue

  if guess_letter in display:
    print(f"You've already guessed the letter {guess_letter}")
    continue

  if guess_letter not in chosen_word:
      print(f"You guessed {guess_letter}, that's not in the word. You lose a life.")
      lives -= 1
      if lives == 0:
        end_of_game = True
        print("😔You lose!")

  if guess_letter in chosen_word:
    for position in range(word_length):
      letter = chosen_word[position]
      if letter == guess_letter:
        display[position] = letter
    print(display)

  if "_" not in display:
    end_of_game = True
    print(f"You guessed the word {chosen_word} correct, 🎉You won!")

  print(stages[lives])
```
