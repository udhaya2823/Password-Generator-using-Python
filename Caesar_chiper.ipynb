```python
#caesar_chiper
alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n',
            'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z',]

def process(text, shift, what_to_do):
  chiper_text = ""
  for letter in text:
    if letter not in alphabet:
      chiper_text += letter
    else:
      position = alphabet.index(letter)
      if what_to_do == "decrypt":
        new_position = (position - shift) % 26
        chiper_text += alphabet[new_position]
      else:
        new_position = (position + shift) % 26
        chiper_text += alphabet[new_position]
  print(f"The encoded text is {chiper_text}")

flag = True
while flag:
  what_to_do = input("Type 'encrypt' for Encryption, Type 'decrypt' for Decryption\n").lower()
  text = input("Type your message:\n").lower()
  shift = int(input("Type the shift number:\n"))
  if what_to_do == "encrypt" or "decrypt":
      process(text, shift, what_to_do)
  else:
      print("Invalid input")
  play_again = input("Type 'YES' if you want to go again. Otherwise type 'NO'\n").lower()
  if play_again == "no":
    flag = False
    print("Goodbye")
```
